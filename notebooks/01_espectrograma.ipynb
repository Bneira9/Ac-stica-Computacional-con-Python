```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

# 1. Cargar un audio de ejemplo que viene incluido en librosa
# (Esto simula que cargamos nuestra primera canción)
ruta_audio = librosa.ex('trumpet') 
y, sr = librosa.load(ruta_audio)

print(f"Audio cargado exitosamente.")
print(f"Frecuencia de muestreo (sr): {sr} Hz")
print(f"Duración: {len(y)/sr:.2f} segundos")

# 2. Calcular el espectrograma (convertir el sonido en una imagen de frecuencias)
D = librosa.stft(y)  # STFT: Short-Time Fourier Transform
S_db = librosa.amplitude_to_db(np.abs(D), ref=np.max)

# 3. Dibujar el espectrograma
plt.figure(figsize=(10, 5))
librosa.display.specshow(S_db, sr=sr, x_axis='time', y_axis='hz')
plt.colorbar(format='%+2.0f dB')
plt.title('Espectrograma de Frecuencias (Audio de Prueba)')
plt.tight_layout()
plt.show()
```
